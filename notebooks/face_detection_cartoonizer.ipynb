```python
model = YOLO(r"runs\detect\train8\weights\best.pt")
results = model([r"messi.jpg"],iou=0)  
```

```python
results[0].show()

for result in results:
    for box in result.boxes.xyxy: 
        x1, y1, x2, y2 = map(int, box[:4])
        
     
        image = cv2.imread("messi.jpg")
        face_crop = image[y1:y2, x1:x2]

cv2.imshow("cropped face",face_crop)
cv2.waitKey(0)
```

```python


gray = cv2.cvtColor(face_crop , cv2.COLOR_BGR2GRAY)


gray = cv2.medianBlur(gray, 5)


masks = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, 
                              cv2.THRESH_BINARY, 11, 3)


smoothed = cv2.bilateralFilter(face_crop, 9, 75,75)


cartoon = cv2.bitwise_and(smoothed, smoothed, mask=masks)


cv2.imshow("Cartoonized Image", cartoon)
cv2.waitKey(0)

```
